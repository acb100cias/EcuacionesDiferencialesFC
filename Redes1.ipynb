```python
N=list(range(101))
```

```python
E=[(np.random.choice(N),np.random.choice(N)) for i in range(300)]
```

```python
G1=nw.Graph()
G1.add_nodes_from(N)
G1.add_edges_from(E)
```

```python
nw.draw(G1)
```
