```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate as itg
```

```python
def OA(Y,t):
  v,y=Y
  dy=v
  dv=-b/m*v-k/m*y
  Y=[dv,dy]
  return Y
```

# Sin amortiguamiento

$b=0$



```python
b=0
k=1
m=1
y0=[1,0]
```

```python
y0=[1,0]
T=np.linspace(0,10,100)

sol=itg.odeint(OA,y0,T).T
```

```python
plt.plot(T,sol[0])
plt.plot(T,sol[1])
```

```python
plt.plot(sol[1],sol[0])
#
```

```python
y0=[0,0]
T=np.linspace(0,10,100)
for i in range(1,5):
  sol=itg.odeint(OA,y0,T).T
  plt.plot(sol[1],sol[0])
  y0[0]+=1/i
#
```
