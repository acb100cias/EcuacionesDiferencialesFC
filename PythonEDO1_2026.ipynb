#Método de Euler

Aproximación numérica a la solución de la ecuación

$$
\frac{\mathrm{d}x}{\mathrm{d}t}=f(t,x)
$$

con condición inicial

$$ x(0)=x_0$$

Escribimos la ecuación para la recta tangente a la solución por $(t_0,x_0)$ como:

$$
x=x_0+f(t_0,x_0)(t-t_0)
$$

esta expresión nos dá un segundo punto.

¿Cómo lo usamos para aproximar la solución?

El primer paso es determinar una partición $h$-regular del intervalo $[t_0,t_f]$, tomamos $t_{i+1}=t_i+h$ y consideramos la colección dada por:

$$
x_{n+1}=x_n+f(t_n,x_n)(t_{n+1}-t_n)=x_n+f(t_n,x_n)h
$$



image.png


```python
def f(x):
  return r*x
```

```python
r=0.1
T=np.arange(0.1,100,0.01)
h=0.01
X=[f(T[0])]
```

```python
for i in T[1:]:
  x=X[-1]+f(X[-1])*h
  X.append(x)
```

```python
X
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(T,X,'--',label='Solución numérica')

plt.plot(T,f(T[0])*np.exp(r*T),'.',alpha=0.3,label='Solución exacta')
plt.legend()
```

```python
def g(x):
  return r*x*(1-(x/k))
k=10
T=np.arange(0.1,200,0.01)
X=[g(T[0])]
for i in T[1:]:
  x=X[-1]+g(X[-1])*h
  X.append(x)
plt.plot(T,X,'--',label='Solución numérica')
plt.plot(T,np.ones(len(T))*k,label='solución estacionaria')

```


#Método de Runge-Kutta

$$
x_0=x(0)
$$

y tomando una $h$-partición del intervalo $[t_0,t_f]$

$$
x_{n+1}=x_n+\frac{1}{6}(k_1+2k_2+2k_3+k_4)
$$

donde

$$
k_1=hf(t_n,x_n)
$$

$$
k_2=hf(t_n + \frac{1}{2}h, x_n + k_1)
$$


$$
k_3=hf(t_n + (\frac{1}{2})h, x_n + (\frac{1}{2})k_2)
$$

$$
k_4=hf(t_n + h, x_n + k3)
