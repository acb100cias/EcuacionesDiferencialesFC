```python
from numpy import *
from matplotlib.pyplot import *
```

```python
dt=0.1

def inicializa():
  global x, xresult,y ,yresult
  x=y=0.1
  xresult=[x]
  yresult=[y]

def actualiza():
  global x,xresult,y,yresult
  nextx=x+y*dt
  nexty=y+(-mu*(x**2-1)*y-x)*dt
  x,y=nextx,nexty

def observa():
  global x,xresult,y,yresult
  xresult.append(x)
  yresult.append(y)
```

```python
def planofase():
  inicializa()
  for t in range(10000):
    actualiza()
    observa()
  plot(xresult,yresult,c='b')
  axis('image')
  axis([-10,10,-10,10])
  title(r'$\mu$='+str(mu))
```

```python
mus=[-2,-1,-0.5,0,0.5,1,1.5,2]
for i in range(len(mus)):
  mu=mus[i]
  planofase()
  show()
```
